# Data Statistics of MedMCQA Dataset

```python
from jsonlines import jsonlines
import json
import prettytable
```

```python
with jsonlines.open("../data/MedMCQA/original/dev.jsonl", "r") as f:
    medmcqa_dev = [entry for entry in f]

with jsonlines.open("../data/MedMCQA/original/test.jsonl", "r") as f:
    medmcqa_test = [entry for entry in f]

with jsonlines.open("../data/MedMCQA/original/train.jsonl", "r") as f:
    medmcqa_train = [entry for entry in f]
```

```python
print(json.dumps(medmcqa_dev[79], indent=4))
```

```python
medmcqa_dev[0]["exp"] is None
```

```python
# not necessary for our task
topics = [entry["topic_name"] for entry in medmcqa_dev if "topic_name" in entry and entry["topic_name"]]
print("Number of topics in DEV:", len(set(topics)))
for topic_name in set(topics):
    print(" -", topic_name)
```

```python
choice = [entry["choice_type"] for entry in medmcqa_dev if "choice_type" in entry and entry["choice_type"]]
print("Number of topics in DEV:", len(set(choice)))
for c in set(choice):
    print(" -", c)
```

```python
subjects = [entry["subject_name"] for entry in medmcqa_dev if "subject_name" in entry and entry["subject_name"]]
print("Number of subjects in DEV:", len(set(subjects)))
for subject_name in set(subjects):
    print(" -", subject_name, ":", subjects.count(subject_name))

no_subject = [entry for entry in medmcqa_dev if not ("subject_name" in entry and entry["subject_name"])]
print("Number of questions in DEV without a subject:", len(no_subject))
```

```python
def exp(entry):
    if "exp" in entry and entry["exp"]:
        return True
    else:
        return False
```

```python
filtered_dev = [entry for entry in medmcqa_dev if exp(entry)]
filtered_test = [entry for entry in medmcqa_test if exp(entry)]
filtered_train = [entry for entry in medmcqa_train if exp(entry)]
```

```python
print(f"Total number of cases in DEV {len(medmcqa_dev)}; after filtering:", len(filtered_dev))
print(f"Total number of cases in TEST {len(medmcqa_test)}; after filtering:", len(filtered_test))
print(f"Total number of cases in TRAIN {len(medmcqa_train)}; after filtering:", len(filtered_train))
```
