```python
from datasets import load_dataset
from transformers import XLNetTokenizer
tokenizer = XLNetTokenizer.from_pretrained("xlnet/xlnet-base-cased")
```

## PubMedQA

```python
all_contexts = []
for sample in ds_artificial['train']:
    all_contexts.extend(sample['context']['contexts'])
```

```python
len(all_contexts)
```

```python
context_tokens = []
for context in all_contexts:
    context_tokens.extend(tokenizer.tokenize(context))
```

```python
set_context_tokens = set(context_tokens)
```

```python
freq_list = {}
for token in context_tokens:
    if token in freq_list:
        freq_list[token] += 1
    else:
        freq_list[token] = 1
```

```python
sorted_freq_list = dict(sorted(freq_list.items(), key=lambda item: item[1], reverse=True))
```

```python
print("Number of tokens in 'train':", len(context_tokens)) # 74 380 798
print("Vocabulary without filtering:", len(set_context_tokens)) # 27 369
```

```python
print("Top 50 tokens:")
for i, (token, freq) in enumerate(sorted_freq_list.items()):
    if i >= 50:
        break
    print(f"  {token}: {freq}")
```
