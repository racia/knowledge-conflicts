```python
import os
import json

conflict_data, unique_points = [], set()

for fn in os.listdir(''):
    print(fn)
    if fn.endswith('.jsonl'):
        with open(fn, 'r') as f:
            if fn.endswith('27024'):
                lines = json.load(f)
            else:
                lines = f.read().strip().split('\n')
                lines = [json.loads(l) for l in lines]

            for line in lines:
                if isinstance(line, list):
                    for l in line:
                        if l['question'] not in unique_points:
                            unique_points.add(l['question'])
                            conflict_data.append(l)
                else:
                    if line['question'] not in unique_points:
                        unique_points.add(line['question'])
                        conflict_data.append(line)

print(len(conflict_data))

with open('train_conflict.json', 'w') as fw:
    json.dump(conflict_data, fw, indent=2)
```
