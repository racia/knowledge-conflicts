```python
import json
```

```python

with open('paraphrases/dev-paraphrases.jsonl', 'r') as f:
     lines = f.read().strip().split('\n')
     paraphrases = [json.loads(line) for line in lines]

     min_opt = float('inf')
     missing_paraphrases = 0
     for paraphrase in paraphrases:
         if 'question_op' not in paraphrase:
             missing_paraphrases += 1
         else:
             if min_opt > len(paraphrase['question_op']):
                 min_opt = len(paraphrase['question_op'])

## we have 886 missing paraphrases and questions with only 1 paraphrase
## on inference, we would like to choose the best paraphrase for each question
print(min_opt, missing_paraphrases)
```
