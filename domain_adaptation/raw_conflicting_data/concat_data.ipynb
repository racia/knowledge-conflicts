```python
import os
import json

conflict_data = []
for fn in os.listdir('.'):
    if fn.endswith('.jsonl'):
        print(fn)
        with open(fn, 'r') as f:
            lines = f.read().strip().split('\n')
            conflict_data.extend([json.loads(line) for line in lines])

print(len(conflict_data))

with open('train_conflict.json', 'w') as fw:
    json.dump(conflict_data, fw, indent=2)
```
