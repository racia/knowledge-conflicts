```python
import torch
from transformers import MambaForCausalLM, AutoTokenizer, BitsAndBytesConfig
```

```python
model_name = "state-spaces/mamba-1.4b-hf"
device = torch.device("cuda" if torch.cuda.is_available() else "mps")
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    llm_int8_enable_fp32_cpu_offload=True,
)
model = MambaForCausalLM.from_pretrained("state-spaces/mamba-1.4b-hf").to(device)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
input_ids = tokenizer("Hey how are you doing?", return_tensors="pt")["input_ids"]
print("inputs", tokenizer.decode(input_ids, skip_special_tokens=False))
outputs = model.generate(intut_ids=input_ids, max_new_tokens=100)
print(tokenizer.decode(outputs[0], skip_special_tokens=False))
```
